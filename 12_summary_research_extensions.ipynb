## Limitations of Visium
- **Multi-cell spots** (~1-10 cells): not single-cell resolution; needs deconvolution for
  composition. (Newer platforms - Visium HD, Xenium, MERFISH - push toward single-cell/
  subcellular resolution.)
- **Capture efficiency**: only a fraction of transcripts are captured; zeros are ambiguous.
- **2D single section**: one slice of a 3D organ; batch/stain effects across slides.
- **Registration is approximate** at the cell scale.


## How a pharma team would extend this
- **Treated vs untreated** - run the same pipeline on both arms and compare domain
  composition and expression to localize drug effects (spatial pharmacodynamics).
- **Tumor microenvironment remodeling** - quantify how tumor/immune/stromal domains
  rearrange with therapy.
- **Immune exclusion** - measure distance from tumor to immune-infiltrated domains as a
  response biomarker.
- **Spatial pharmacodynamic biomarkers** - find spatially structured genes that change on
  treatment and are readable from histology.
- **Toxicity profiling** - localize injury/stress programs to specific tissue zones in
  safety studies.
- **Fibrosis / inflammation mapping** - score and track ECM/immune programs spatially.
- **Target expression in niches** - confirm a target is expressed in the intended
  compartment, not just on bulk average.
- **Histology->transcriptomics prediction** - scale up notebook 10 across cohorts to
  predict molecular state from cheap H&E (with stain normalization + proper validation).
- **Multimodal foundation models** - combine pathology image encoders with expression
  models for spatially-aware, transferable representations.

## What this means biologically (the big picture)
Tissue is organized, and organization drives biology and disease. Spatial transcriptomics
lets us read molecular programs *in place* and tie them to the morphology clinicians
already use. For drug discovery, that means finding where a target lives, where a drug
acts, and how the tissue ecosystem responds - questions bulk and dissociated assays
cannot answer.
